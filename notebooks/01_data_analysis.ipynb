```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

path = Path.home() / "ml-lab/data/combined_dataset.csv"

df = pd.read_csv(path)

counts = df["label"].value_counts()

print("Количество объектов:")
print(counts)

print("\nДоли классов, %:")
print((counts / counts.sum() * 100).round(2))

counts.plot(kind="bar", rot=0, title="Распределение классов")
plt.ylabel("Количество объектов")
plt.tight_layout()
plt.show()

print(f"Разница между классами: {counts.max() - counts.min():,}")
```
