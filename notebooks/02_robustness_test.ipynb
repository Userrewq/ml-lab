```python
import pandas as pd
import numpy as np
import sys
import sys
sys.path.append('../src')
from val_predict import DNSPredictor

X_test_20 = pd.read_csv('../data/X_test_v1.csv').head(20)
y_test_20 = pd.read_csv('../data/y_test_v1.csv').head(20).values

predictor = DNSPredictor(model_path='../models/logistic_regression_model.onnx')
```

```python
preds_orig = predictor.predict(X_test_20)
```

```python
X_rec_mod = X_test_20.copy()
X_rec_mod['distinct_A_records'] = X_rec_mod['distinct_A_records'] * 2
preds_rec = predictor.predict(X_rec_mod)

changed_rec = sum(1 for o, m in zip(preds_orig, preds_rec) if o != m)
print(f"\nЭксперимент 1: Увеличение distinct_A_records в 2 раза")
print(f"Изменилось предсказаний: {changed_rec} из 20 ({changed_rec/20*100:.1f}%)")
print("Отличия в предсказаниях\n")
for i in range(20):
    if preds_orig[i] != preds_rec[i]:
        print(f"Строка {i}: Истинная метка = {y_test_20[i]}, "
              f"Оригинал = {preds_orig[i]}, Модифицированный = {preds_rec[i]}")
```

```python
X_port_mod = X_test_20.copy()
X_port_mod['src_port'] = 99999
preds_port = predictor.predict(X_port_mod)

changed_port = sum(1 for o, m in zip(preds_orig, preds_port) if o != m)

print(f"\nЭксперимент 2: выброс в значение src_port")
print(f"Изменилось предсказаний: {changed_port} из 20 ({changed_port/20*100:.1f}%)")
print("Отличия в предсказаниях\n")
for i in range(20):
    if preds_orig[i] != preds_port[i]:
        print(f"Строка {i}: Истинная метка = {y_test_20[i]}, "
              f"Оригинал = {preds_orig[i]}, Модифицированный = {preds_port[i]}")
```
